## Part B: SQL Analysis

### Chapter 1: How Has the Tournament Grown?

**B1: Number of teams, matches, and total goals per tournament**

```python
tournament_grow = pd.read_sql("""
                               SELECT year, count_teams, total_matches, total_goals
                               FROM tournaments
                               LEFT JOIN (
                                   SELECT tournament_id, COUNT(*) AS total_matches
                                   FROM matches
                                   GROUP BY tournament_id
                                   ) AS match_counts
                                USING(tournament_id)
                                LEFT JOIN (
                                    SELECT tournament_id, COUNT(*) AS total_goals
                                    FROM goals
                                    GROUP BY tournament_id
                                    ) AS goal_counts
                                USING(tournament_id)
                                WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                ORDER BY year """, conn)

tournament_grow                        
```

The first early World Cup era (1930–1978) stayed small around 13–16 teams at the beginning playing only 18 matches. As the years passed, the modern football era took place, the participating teams kept growing up to 32 teams in the year of 2022 with 64 matches played. Regarding the total goals scored, we cannot skip the 1954 World Cup with the remarkable 140 goals scored in only 26 matches — the highest goals-per-match rate in World Cup history. Similar total goals jumps can also be noticed in later years: 1982 with 146 goals and 1998 with 171 goals, both coinciding with tournament expansions (16→24 teams and 24→32 teams respectively).

**B2: Unique countries by decade — how has the World Cup's global reach expanded?**

```python
unique_countries_total = pd.read_sql("""
                                        SELECT COUNT(DISTINCT team_name) AS unique_countries
                                        FROM team_appearances
                                        WHERE tournament_name LIKE '%FIFA Men''s World Cup' """, conn)
unique_countries_total
```

```python
unique_countries_decade = pd.read_sql("""
                                        SELECT SUBSTR(tournament_name, 1, 3) || '0s' AS decade,
                                        COUNT(DISTINCT team_name) AS unique_countries
                                        FROM team_appearances
                                        WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                        GROUP BY decade
                                        ORDER BY decade ASC """, conn)
unique_countries_decade
```

A total of 85 unique countries have participated in the men's World Cup across its 92-year history. The 1940s are absent from the data due to World War II — no tournaments were held in 1942 or 1946. The biggest jumps in global reach came in the 1970s (32 countries) and 1990s (45 countries), reflecting both tournament expansions and the growing worldwide popularity of football. The 2020s show only 32 countries, but this represents a single tournament (Qatar 2022) rather than a decline.

### Chapter 2: How Has Scoring Changed?

**B4: Average goals per match per tournament — the signature curve of football analytics**

```python
tournament_grow['goals_per_match'] = round(tournament_grow['total_goals'] / tournament_grow['total_matches'], 2)
tournament_grow
```

The 1954 World Cup stands as both the peak and the outlier at 5.38 goals per match — nearly double any modern tournament. From that point on, the average settled around 2.5 goals per match, reflecting a shift toward more tactical, defensively organized football. The modern 32-team era (1998–2022) shows remarkable stability, hovering between 2.23 and 2.69 goals per match despite nearly doubling the number of games — a clear sign that the style of play, not the volume of matches, drives the average.

**B5: Goal breakdown by type — are modern World Cups more penalty-dependent?**

```python
goal_breakdown = pd.read_sql("""
                                SELECT SUBSTR(tournament_name, 1, 4) AS year, SUM(penalty) AS penalty_goals, SUM(own_goal) AS own_goals, 
                                COUNT(*) - SUM(penalty) - SUM(own_goal) AS open_play_goals
                                FROM goals
                                WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                GROUP BY year
                                ORDER BY year
                                """, conn)
goal_breakdown
```

```python
goal_breakdown['penalty_pct'] = round(goal_breakdown['penalty_goals'] * 100 / (goal_breakdown['penalty_goals'] 
                                + goal_breakdown['own_goals'] + goal_breakdown['open_play_goals']), 1)
goal_breakdown
```

Penalty goals have grown in raw numbers over the decades, but much of that increase is explained by more matches being played — the percentage has hovered between 5–11% for most tournaments. The clear outlier is 2018 (13%), which coincided with the introduction of VAR technology, awarding penalties that referees might previously have missed. That same year also saw a record 12 own goals, though this is likely coincidental rather than VAR-related.

**B6: Goal timing distribution — when do World Cup goals happen?**

```python
goal_timing = pd.read_sql("""
                            SELECT 
                                CASE 
                                    WHEN minute_regulation <= 15 THEN '1-15'
                                    WHEN minute_regulation <= 30 THEN '16-30'
                                    WHEN minute_regulation <= 45 THEN '31-45'
                                    WHEN minute_regulation <= 60 THEN '46-60'
                                    WHEN minute_regulation <= 75 THEN '61-75'
                                    WHEN minute_regulation <= 90 THEN '76-90'
                                    WHEN minute_regulation > 90 THEN '90+'
                                END AS minute_bucket,
                                COUNT (*) AS total_goals
                            FROM goals
                            WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                            GROUP BY minute_bucket
                            """, conn)
goal_timing
```

Goals steadily increase as the match progresses, with the 76–90 minute period being the most dangerous (579 goals — nearly double the 1–15 minute bucket). Overall, second halves produce roughly 30% more goals than first halves, reflecting tired defenders, tactical substitutions, and the urgency of chasing a result late in the game.

**B7: Blowout analysis — do expanded tournaments produce more lopsided matches?**

```python
blowout_analysis = pd.read_sql("""
                                SELECT SUBSTR(tournament_name, 1, 4) AS year,
                                SUM(CASE WHEN ABS(home_team_score - away_team_score) >= 3 THEN 1 ELSE 0 END) AS blowouts,
                                COUNT(*) AS total_matches
                                FROM matches
                                WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                GROUP BY year
                                ORDER BY year
                                """, conn)
blowout_analysis['blowout_pct'] = round(blowout_analysis['blowouts'] * 100.0 / blowout_analysis['total_matches'], 1)
blowout_analysis
```

Expanded tournaments did not produce more blowouts — the opposite, in fact. The early era was far more lopsided, with 1930 (55.6%) and 1954 (42.3%) leading by a wide margin. The modern 32-team era has stabilized between 10–19%, and the 1998 expansion from 24 to 32 teams barely changed the blowout rate. This suggests that the qualifying process has kept the overall standard competitive, even as more teams enter the tournament.
