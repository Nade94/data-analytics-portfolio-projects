## Part B: SQL Analysis

### Chapter 1: How Has the Tournament Grown?

**B1: Number of teams, matches, and total goals per tournament**

```python
tournament_grow = pd.read_sql("""
                               SELECT year, count_teams, total_matches, total_goals
                               FROM tournaments
                               LEFT JOIN (
                                   SELECT tournament_id, COUNT(*) AS total_matches
                                   FROM matches
                                   GROUP BY tournament_id
                                   ) AS match_counts
                                USING(tournament_id)
                                LEFT JOIN (
                                    SELECT tournament_id, COUNT(*) AS total_goals
                                    FROM goals
                                    GROUP BY tournament_id
                                    ) AS goal_counts
                                USING(tournament_id)
                                WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                ORDER BY year """, conn)

tournament_grow                        
```

The first early World Cup era (1930–1978) stayed small around 13–16 teams at the beginning playing only 18 matches. As the years passed, the modern football era took place, the participating teams kept growing up to 32 teams in the year of 2022 with 64 matches played. Regarding the total goals scored, we cannot skip the 1954 World Cup with the remarkable 140 goals scored in only 26 matches — the highest goals-per-match rate in World Cup history. Similar total goals jumps can also be noticed in later years: 1982 with 146 goals and 1998 with 171 goals, both coinciding with tournament expansions (16→24 teams and 24→32 teams respectively).

**B2: Unique countries by decade — how has the World Cup's global reach expanded?**

```python
unique_countries_total = pd.read_sql("""
                                        SELECT COUNT(DISTINCT team_name) AS unique_countries
                                        FROM team_appearances
                                        WHERE tournament_name LIKE '%FIFA Men''s World Cup' """, conn)
unique_countries_total
```

```python
unique_countries_decade = pd.read_sql("""
                                        SELECT SUBSTR(tournament_name, 1, 3) || '0s' AS decade,
                                        COUNT(DISTINCT team_name) AS unique_countries
                                        FROM team_appearances
                                        WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                        GROUP BY decade
                                        ORDER BY decade ASC """, conn)
unique_countries_decade
```

A total of 85 unique countries have participated in the men's World Cup across its 92-year history. The 1940s are absent from the data due to World War II — no tournaments were held in 1942 or 1946. The biggest jumps in global reach came in the 1970s (32 countries) and 1990s (45 countries), reflecting both tournament expansions and the growing worldwide popularity of football. The 2020s show only 32 countries, but this represents a single tournament (Qatar 2022) rather than a decline.

### Chapter 2: How Has Scoring Changed?

**B4: Average goals per match per tournament — the signature curve of football analytics**

```python
tournament_grow['goals_per_match'] = round(tournament_grow['total_goals'] / tournament_grow['total_matches'], 2)
tournament_grow
```

The 1954 World Cup stands as both the peak and the outlier at 5.38 goals per match — nearly double any modern tournament. From that point on, the average settled around 2.5 goals per match, reflecting a shift toward more tactical, defensively organized football. The modern 32-team era (1998–2022) shows remarkable stability, hovering between 2.23 and 2.69 goals per match despite nearly doubling the number of games — a clear sign that the style of play, not the volume of matches, drives the average.

**B5: Goal breakdown by type — are modern World Cups more penalty-dependent?**

```python
goal_breakdown = pd.read_sql("""
                                SELECT SUBSTR(tournament_name, 1, 4) AS year, SUM(penalty) AS penalty_goals, SUM(own_goal) AS own_goals, 
                                COUNT(*) - SUM(penalty) - SUM(own_goal) AS open_play_goals
                                FROM goals
                                WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                GROUP BY year
                                ORDER BY year
                                """, conn)
goal_breakdown
```

```python
goal_breakdown['penalty_pct'] = round(goal_breakdown['penalty_goals'] * 100 / (goal_breakdown['penalty_goals'] 
                                + goal_breakdown['own_goals'] + goal_breakdown['open_play_goals']), 1)
goal_breakdown
```

Penalty goals have grown in raw numbers over the decades, but much of that increase is explained by more matches being played — the percentage has hovered between 5–11% for most tournaments. The clear outlier is 2018 (13%), which coincided with the introduction of VAR technology, awarding penalties that referees might previously have missed. That same year also saw a record 12 own goals, though this is likely coincidental rather than VAR-related.

**B6: Goal timing distribution — when do World Cup goals happen?**

```python
goal_timing = pd.read_sql("""
                            SELECT 
                                CASE 
                                    WHEN minute_regulation <= 15 THEN '1-15'
                                    WHEN minute_regulation <= 30 THEN '16-30'
                                    WHEN minute_regulation <= 45 THEN '31-45'
                                    WHEN minute_regulation <= 60 THEN '46-60'
                                    WHEN minute_regulation <= 75 THEN '61-75'
                                    WHEN minute_regulation <= 90 THEN '76-90'
                                    WHEN minute_regulation > 90 THEN '90+'
                                END AS minute_bucket,
                                COUNT (*) AS total_goals
                            FROM goals
                            WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                            GROUP BY minute_bucket
                            """, conn)
goal_timing
```

Goals steadily increase as the match progresses, with the 76–90 minute period being the most dangerous (579 goals — significantly more than the 1–15 minute bucket). Overall, second halves produce roughly 30% more goals than first halves, reflecting tired defenders, tactical substitutions, and the urgency of chasing a result late in the game.

**B7: Blowout analysis — do expanded tournaments produce more lopsided matches?**

```python
blowout_analysis = pd.read_sql("""
                                SELECT SUBSTR(tournament_name, 1, 4) AS year,
                                SUM(CASE WHEN ABS(home_team_score - away_team_score) >= 3 THEN 1 ELSE 0 END) AS blowouts,
                                COUNT(*) AS total_matches
                                FROM matches
                                WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                GROUP BY year
                                ORDER BY year
                                """, conn)
blowout_analysis['blowout_pct'] = round(blowout_analysis['blowouts'] * 100.0 / blowout_analysis['total_matches'], 1)
blowout_analysis
```

Expanded tournaments did not produce more blowouts — the opposite, in fact. The early era was far more lopsided, with 1930 (55.6%) and 1954 (42.3%) leading by a wide margin. The modern 32-team era has stabilized between 10–19%, and the 1998 expansion from 24 to 32 teams barely changed the blowout rate. This suggests that the qualifying process has kept the overall standard competitive, even as more teams enter the tournament.

### Chapter 3: How Has Discipline Evolved?

**B8: Cards per match per tournament (1970–2022)**

```python
cards_per_match = pd.read_sql("""
                                SELECT tournament_name, ROUND(yellow_cards * 1.0 / total_matches, 2) AS yellow_per_match, 
                                ROUND(red_cards * 1.0 / total_matches, 2) AS red_per_match
                                FROM (
                                    SELECT tournament_name, SUM(yellow_card) AS yellow_cards, SUM(red_card) AS red_cards
                                    FROM bookings
                                    GROUP BY tournament_name) AS card_counts
                                INNER JOIN (
                                    SELECT tournament_name, COUNT(match_name) AS total_matches
                                    FROM matches
                                    GROUP BY tournament_name) AS match_count
                                USING(tournament_name)
                                WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                ORDER BY tournament_name """, conn)
cards_per_match                 
```

Yellow card rates climbed steadily from 1.63 per match in 1970 to a peak of 5.09 in 2006 — the roughest World Cup by card volume — before declining to around 3.5 in the most recent tournaments. Red cards have always been far less frequent, peaking at 0.28 per match in 1998 and dropping sharply in the VAR era (0.03 in 2018, 0.02 in 2022), suggesting that video review may be deterring the most reckless fouls rather than increasing punishment.

**B9: Cards by tournament stage**

```python
cards_by_stage = pd.read_sql("""
                                SELECT stage_name, ROUND(yellow_cards * 1.0 / total_matches, 2) AS yellow_per_match, 
                                ROUND(red_cards * 1.0 / total_matches, 2) AS red_per_match
                                FROM (
                                    SELECT stage_name, SUM(yellow_card) AS yellow_cards, SUM(red_card) AS red_cards
                                    FROM bookings
                                    WHERE tournament_name LIKE '%FIFA Men''s World Cup' AND tournament_name LIKE '20%'
                                    GROUP BY stage_name) AS card_counts
                                INNER JOIN (
                                    SELECT stage_name, COUNT(match_name) AS total_matches
                                    FROM matches
                                    WHERE tournament_name LIKE '%FIFA Men''s World Cup' AND tournament_name LIKE '20%'
                                    GROUP BY stage_name) AS match_count
                                USING(stage_name)
                                ORDER BY yellow_per_match DESC """, conn)
cards_by_stage
```

Card rates rise through the early knockout rounds — from 3.77 yellows per match in the group stage to 4.79 in the quarter-finals — before dropping sharply to 2.75 in the semi-finals and spiking again to 5.67 in the final. However, sample sizes vary dramatically: the group stage covers 288 matches across 2002–2022, while semi-finals and finals contribute only 12 and 6 matches respectively. With samples that small, a single rough final can skew the average, so the quarter-final rate is the most reliable measure of knockout intensity.

**B10: Pre-VAR vs post-VAR comparison**

```python
var_impact = pd.read_sql("""
                           SELECT era, ROUND(yellow_cards * 1.0 / total_matches, 2) AS yellow_per_match, 
                                ROUND(red_cards * 1.0 / total_matches, 2) AS red_per_match
                           FROM (
                                SELECT 
                                    CASE 
                                        WHEN tournament_name LIKE '2010%' OR tournament_name LIKE '2014%' THEN 'pre-VAR'
                                        WHEN tournament_name LIKE '2018%' OR tournament_name LIKE '2022%' THEN 'post-VAR'
                                    END AS era, 
                                    SUM(yellow_card) AS yellow_cards, SUM(red_card) AS red_cards
                                FROM bookings
                                WHERE tournament_name LIKE '%FIFA Men''s World Cup' AND (tournament_name LIKE '201%' 
                                OR tournament_name LIKE '202%')
                                GROUP BY era) AS card_counts
                            INNER JOIN (
                                        SELECT
                                            CASE 
                                                WHEN tournament_name LIKE '2010%' OR tournament_name LIKE '2014%' THEN 'pre-VAR'
                                                WHEN tournament_name LIKE '2018%' OR tournament_name LIKE '2022%' THEN 'post-VAR'
                                            END AS era, COUNT(match_name) AS total_matches
                                    FROM matches
                                    WHERE tournament_name LIKE '%FIFA Men''s World Cup' AND (tournament_name LIKE '201%' 
                                    OR tournament_name LIKE '202%')
                                    GROUP BY era) AS match_count
                            USING(era) """, conn)
var_impact                     
```

Yellow card rates barely changed between the pre-VAR (3.42 per match) and post-VAR era (3.48) — the everyday foul economy remained the same. Red cards, however, dropped 85%, from 0.13 to 0.02 per match. Two forces likely explain this: players know that every challenge is being reviewed from multiple angles, which deters the reckless tackles that earn reds in the first place; and when a referee does show red, VAR can intervene to overturn or downgrade the decision if replays suggest the foul didn't warrant maximum punishment.

```python
debut_nations = pd.read_sql("""
                                SELECT SUBSTR(debut, 1, 4) AS year, COUNT(*) AS debut_nations
                                FROM (
                                      SELECT team_name, MIN(tournament_name) AS debut
                                      FROM team_appearances
                                      WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                                      GROUP BY team_name)
                                WHERE debut IN ('1994 FIFA Men''s World Cup', '1998 FIFA Men''s World Cup')
                                GROUP BY year
                                """, conn)
debut_nations
```

```python
expansion_comparison = pd.DataFrame({
    'year': [1994, 1998],
    'goals_per_match': [2.71, 2.67],
    'blowout_pct': [15.4, 18.8],
    'yellow_per_match': [4.38, 3.97],
    'red_per_match': [0.15, 0.28],
    'draws_pct': [5.8, 6.2],
    'debut_nations': [4, 4]
})

expansion_comparison
```

The 24→32 expansion had surprisingly little impact on the quality of play. Goals per match barely moved (2.71 → 2.67) and 0-0 draws stayed flat (5.8% → 6.2%), suggesting the extra teams didn't dilute attacking football. Blowouts rose modestly (15.4% → 18.8%), hinting that a few newcomers faced steep mismatches. The discipline picture was mixed: yellow cards per match actually decreased (4.38 → 3.97) while reds nearly doubled (0.15 → 0.28), pointing to fewer routine fouls but more extreme ones. Both tournaments introduced 4 debut nations. For 2026's 32→48 expansion, this precedent is cautiously optimistic — the last expansion preserved the core product even as it opened the door wider.

**B12: Top teams all-time performance**

```python
top_teams = pd.read_sql("""
                            SELECT CASE 
                                    WHEN team_name = 'West Germany' THEN 'Germany' 
                                    ELSE team_name
                                    END AS team_name_cleared,
                            COUNT(*) AS total_matches, SUM(win) AS wins, SUM(draw) AS draws,
                            SUM(lose) AS loses, SUM(goals_for) AS goals_for, SUM(goals_against) AS goals_against,
                            ROUND(SUM(win) * 100.0 / COUNT(*), 1) AS wins_pct 
                            FROM team_appearances
                            WHERE tournament_name LIKE '%FIFA Men''s World Cup'
                            GROUP BY team_name_cleared
                            HAVING COUNT(*) >= 15
                            ORDER BY wins_pct DESC
                            LIMIT 10
                            """, conn)

top_teams
```

The World Cup's power structure is unmistakable: Brazil (69.3%) and Germany (64.3%) sit in a tier of their own, with over 110 matches each and win rates well above the rest. Argentina (60.2%) anchors a second tier, followed by a cluster of European nations — Croatia, Netherlands, France, Italy — hovering between 51% and 57%. The most striking entry is Croatia: 56.7% from just 30 matches, a small nation that has only competed since 1998 yet consistently outperforms far larger footballing countries. The continental split tells its own story — 7 of the top 10 are European, 2 South American, with no African, Asian, or North American team making the cut despite decades of participation.
